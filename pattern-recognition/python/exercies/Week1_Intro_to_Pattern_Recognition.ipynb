Loading Data

```python
import pandas as pd
```

```python
df = pd.read_csv("data/Iris.csv")
# Dataset is now stored in a Pandas Dataframe
```

```python
df['Species_Label'] =0
df.loc[df['Species'] == 'Iris-versicolor','Species_Label']=1
df.loc[df['Species'] == 'Iris-virginica'	,'Species_Label']=2
```

```python
print(df)
```

**Simple Classification Task**

```python
import numpy as np
#1.Keep Features
#2 Transform from df to numpy using values
X=df.iloc[:,1:5].values
y=df['Species_Label'].values
```

```python
print(X)
```

#  **Check point1: data spliting**
try modifying the code with different test size and random state

```python
#2.Split data into 2 train and test
#The training data is for understanding the pattern
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.3,
                                                    random_state=42)
```

```python
print("Overall data is ",X.shape)
print("The training data is ",X_train.shape)
print("The training data is ",X_test.shape)
```

# **Check point2:**
Choose (Weak) Classifier. Use only mean of 1 value to define the class

```python
y_pred=np.zeros(np.shape(y_test))
```

```python
print(y_pred)
```

```python
def classifier(X_train,X_test):
  y_pred=np.zeros(np.shape(X_test[:,0]))
  y_pred[X_test[:,0]>np.mean(X_train[:,0])]=1
  y_pred[X_test[:,1]>np.mean(X_train[:,1])]=2
  return y_pred
```

```python
Feature1=2
Feature2=3
y_pred=classifier(X_train[:,[Feature1,Feature2]],X_test[:,[Feature1,Feature2]])
```

```python
print(y_pred)
```

```python
np.array([y_test, y_pred]).T
```

```python
accuracy = np.mean(y_pred == y_test)
print(f"Baseline accuracy: {accuracy:.2f}")
```

Check point3:
Visualization ADVANCED

```python
# Create a grid over feature space
import numpy as np
import matplotlib.pyplot as plt

h = 0.05
x_min, x_max = X[:, Feature1].min() - 1.0, X[:, Feature1].max() + 1.0
y_min, y_max = X[:, Feature2].min() - 1.0, X[:, Feature2].max() + 1.0

xx,yy=  np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))

```

```python
new_xtest=np.array([xx.flatten() ,yy.flatten()]).T
```

```python
new_xtest
```

```python
z=classifier(X_train[:,[Feature1,Feature2]],new_xtest)
zz=z.reshape(xx.shape)
```

```python
# Predict based classifer
#zz = np.full(xx.shape, z)
plt.figure()
plt.contourf(xx, yy, zz, alpha=0.3)
plt.scatter(X[:, Feature1], X[:, Feature2], c=y, alpha=0.85)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Region: MEAN (Trivial)")
plt.grid(True)
plt.show()
```

```python

# Plot decision region
plt.figure()
plt.contourf(xx, yy, zz, alpha=0.3)
plt.scatter(X_test[:, Feature1], X_test[:, Feature2], c=y_test, alpha=0.85)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Region:Mean value (Simple)")
plt.grid(True)
plt.show()
```
