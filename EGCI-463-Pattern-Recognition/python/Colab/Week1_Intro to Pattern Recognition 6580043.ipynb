# **Check point2:**
Choose (Weak) Classifier. Use only mean of 1 value to define the class

```python
y_pred=np.zeros(np.shape(y_test))
```

```python
y_pred
```

```python
def classifier(X_train,X_test):
  y_pred=np.zeros(np.shape(X_test[:,0]))
  y_pred[X_test[:,0]>np.mean(X_train[:,0])]=1
  y_pred[X_test[:,1]>np.mean(X_train[:,1])]=2
  return y_pred
```

The given classification is very simple and probably not very accurate.

The model above do this
- default prediction as 0
- if Pepal_length > mean(Pepal_length) -> classify as 1
- if Pepal_width > mean(Petal_width) -> classify as 2

**note** I also try Sepal instead later.

```python
Feature1=0 #2
Feature2=1 #3
y_pred=classifier(X_train[:,[Feature1,Feature2]],X_test[:,[Feature1,Feature2]])
```

```python
print(y_pred)
```

```python
np.array([y_test, y_pred]).T
```

```python
accuracy = np.mean(y_pred == y_test)
print(f"Baseline accuracy: {accuracy:.2f}")
```

After changing the feature columns to Sepal instead of Petal the accuracy drop to only 27% which show that with current classification algorithm Petal are more suitable feature columns.

However this might not be the case for more complex classfication algorithms such as
1. Linear regression
2. Naive Bayes
3. Kernel SVM

```python
# Create a grid over feature space
import numpy as np
import matplotlib.pyplot as plt

h = 0.05
x_min, x_max = X[:, Feature1].min() - 1.0, X[:, Feature1].max() + 1.0
y_min, y_max = X[:, Feature2].min() - 1.0, X[:, Feature2].max() + 1.0

xx,yy=  np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))

```

create scope of graph by min/max value of the feature columns

Then, `np.meshgrid` uses these values to create a grid of points covering the entire area.

```python
new_xtest=np.array([xx.flatten() ,yy.flatten()]).T
```

```python
new_xtest
```

```python
z=classifier(X_train[:,[Feature1,Feature2]],new_xtest)
zz=z.reshape(xx.shape)
```

```python
# Predict based classifer
#zz = np.full(xx.shape, z)
plt.figure()
plt.contourf(xx, yy, zz, alpha=0.3)
plt.scatter(X[:, Feature1], X[:, Feature2], c=y, alpha=0.85)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Region: MEAN (Trivial)")
plt.grid(True)
plt.show()
```
